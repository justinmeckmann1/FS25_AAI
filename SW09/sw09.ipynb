## Entropy <a name="entropy"></a>
Entropy is a measure of *uncertainty* or randomness associated with a random variable. Entropy of a discrete random variable $X$ (with values from the set ${\mathbf X}$):
$$
H(X)=-\sum_{x\in{\mathbf X}}P(X=x)\log_2(P(X=x))
$$
Entropy is measured in bits.

Example:

${\mathbf X}=\{x_1,x_2,x_3,x_4\}$

| $x_i$  | $P(X=x_i)$ |
|-------|---------|
| $x_1$    | $\frac{1}{2}$ |
| $x_2$    | $\frac{1}{4}$ |
| $x_3$    | $\frac{1}{8}$ |
| $x_4$    | $\frac{1}{8}$ |

The entropy $H(X)$ amounts to 1.75 bits.

Problem:

Compute the entropy of the following distributions:
- ${\mathbf X}=\{x_1,x_2,\dots,x_n\}$, $P(X=x_i)=\frac{1}{n}$
- ${\mathbf X}=\{x_1,x_2,\dots,x_n\}$, $P(X=x_i)=\left\{\begin{array}{ll} 1, & i=1 \\ 0, & i\ne 1\end{array}\right.$

### Entropy of a Language <a name="langentropy"></a>
How about the entropy rate of a language $L$ ?

Entropy of a random vector $w_{1:n}=[w_1,w_2,\dots,w_n]$ that ranges over all sequences of words of length $n$ in some language:
$$
H(w_{1:n})=-\sum_{w_{1:n}}P(w_{1:n})\log_2P(w_{1:n})
$$
Corresponding entropy *rate*:
$$
\frac{1}{n}H(w_{1:n})=-\frac{1}{n}\sum_{w_{1:n}}P(w_{1:n})\log_2P(w_{1:n})
$$
Entropy rate of language $L$:
$$
H(L)=-\lim_{n\to\infty}\frac{1}{n}\sum_{w_{1:n}}P(w_{1:n})\log_2P(w_{1:n})
$$
This quantity reflects the amount of information conveyed per word. If the language is regular in certain ways (i.e., both
stationary and ergodic), then
$$
H(L)=-\lim_{n\to\infty}\frac{1}{n}\log_2P(w_{1:n}).
$$
It has been estimated that the entropy rate is typically somewhere below 10 bits/word, with Finnish having one of the highest entropy rates among languages.

### Entropy and Perplexity <a name="entropyperplex"></a>
In fact, the entropy is related to perplexity:
$$
{\rm perplexity}(X)=2^{H(X)}.
$$

## Softmax Function <a name="softmax"></a>
The last layer, and also intermediate layers, of a neural network typically output *scores* or *logits* for the different categories, but not probabilities. This is why often the *softmax function* is used, which transforms a vector of real numbers into a probability distribution.

Given a vector $z=[z_1,z_2,\dots,z_n]$ of $n$ real numbers (often called logits, especially in neural networks), the softmax function $σ(z)$ calculates the probability $P_i$ for each element $z_i$ as follows:
$$
P_i=\frac{\exp(z_i)}{\sum_{j=1}^n\exp(z_j)}
$$
Note that
- all the values are positive between 0 and 1,
- they all add up to exactly 1,

thus forming a valid probability distribution.

## Cross-Entropy <a name="cross_entropy"></a>
The cross-entropy is useful when the actual probability distribution (e.g. of the language $L$), here given by $p(\cdot)$, is unknown, while our model assumes a probability distribution given by $m(\cdot)$. The
cross-entropy is then defined by
$$
H(p,m)=-\lim_{n\to\infty}\frac{1}{n}\sum_{w_{1:n}}p(w_{1:n})\log_2m(w_{1:n}).
$$
Again, if the language is regular in certain ways (i.e., both
stationary and ergodic), then
$$
H(p,m)=-\lim_{n\to\infty}\frac{1}{n}\log_2m(w_{1:n}).
$$
The cross-entropy $H(p,m)$ is an upper bound on the entropy. That is, for any model $m(\cdot)$:
$$
H(L)\le H(p,m).
$$
See [[1], Chap. 3.7](#literature) for more details.

Due to the above inequality, the *cross-entropy loss* is often used in machine learning as the loss function for training neural network models which output logits. With $z_1,\dots,z_C$ the computed logits, $C$ the number of classes, and $t$ the index ot the target class, the cross-entropy loss is computed as
$$
\ell=-\log\frac{\exp(z_t)}{\sum_{c=1}^C\exp(z_c)}.
$$
We arrive at this formulation from the cross-entropy defined above by using for the actual distribution
$p(x_c)=\left\{\begin{array}{ll} 1, & c=t \\ 0, & c\ne t\end{array}\right..$

## Wav2Vec2 <a name="wav_vev"></a>
In the second part of this lecture we look into *wav2vec2*, a leading machine-learning model for the design of ASR systems [[2]](#literature). It is composed of three general components:
- a feature encoder,
- a quantization module,
- and a transformer (discussed in SW10).

A pretrained model with 29 classes (i.e., labels) is available in torchaudio.

```python
# load pytorch wav2vec2 bundle

import torch
import torchaudio

print("torch version:", torch.__version__)
print("torchaudio version:", torchaudio.__version__)

torch.random.manual_seed(0)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:",device)

bundle = torchaudio.pipelines.WAV2VEC2_ASR_BASE_960H
print("sample rate:", bundle.sample_rate)
print("labels:", bundle.get_labels())
```

```python
# create wav2vec2 model

model = bundle.get_model().to(device)
model
```

```python
# load sample waveform

import IPython
from torchaudio.utils import download_asset

SPEECH_FILE = download_asset("tutorial-assets/Lab41-SRI-VOiCES-src-sp0307-ch127535-sg0042.wav")
waveform, sample_rate = torchaudio.load(SPEECH_FILE)
if sample_rate != bundle.sample_rate:
    waveform = torchaudio.functional.resample(waveform, sample_rate, bundle.sample_rate)
waveform = waveform.to(device)

IPython.display.Audio(SPEECH_FILE)
```

```python
# run the ASR

class CTCDecoder(torch.nn.Module):
    def __init__(self, labels, blank=0):
        super().__init__()
        self.labels = labels
        self.blank = blank

    def forward(self, emission: torch.Tensor) -> str:
        """
        Given a sequence emission over labels, get the best path string.
        args:
            emission (tensor): logit tensors. shape `[num_seq, num_label]`

        returns:
            str: resulting transcript
        """
        indices = torch.argmax(emission, dim=-1)    # [num_seq,]
        indices = torch.unique_consecutive(indices, dim=-1)
        indices = [i for i in indices if i != self.blank]
        return "".join([self.labels[i] for i in indices])

decoder = CTCDecoder(labels=bundle.get_labels())

with torch.inference_mode():
    emission, _ = model(waveform)

transcript = decoder(emission[0])

print(transcript)
```
