# 1 - Klassifikation und Segmentierung

#### Single Layer Perceptron

```python
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler

# Punktkoordinaten: [col, row]
X = np.array([
    [198, 252],
    [232, 155],
    [250, 128],
    [289, 384],
    [232, 397],
    [207, 355],
    [419, 328],
    [329, 282],
    [353, 214],
])

# Klassenlabels
y = np.array([0, 0, 0, 1, 1, 1, 2, 2, 2])

# Vorverarbeitung (Standardisierung oft hilfreich für Perzeptron)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Perzeptron trainieren
clf = Perceptron(max_iter=1000, tol=1e-3)
clf.fit(X_scaled, y)

# Vorhersage & Prüfung
y_pred = clf.predict(X_scaled)
print("Korrekt klassifiziert?", np.array_equal(y_pred, y))
```
